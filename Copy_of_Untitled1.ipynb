```python
def calculate_total_price(prices: list[float]) -> float:
    """
    Calculates the total of a list of prices, rounded to two decimal places.

    Args:
        prices: A list of float values representing prices.

    Returns:
        The total sum of prices, rounded to two decimal places.
    """
    total = sum(prices)
    return round(total, 2)

# Example usage:
price_list = [10.55, 23.99, 5.01, 1.20, 0.758]
total_amount = calculate_total_price(price_list)
print(f"Original prices: {price_list}")
print(f"Total price (rounded to 2 decimals): {total_amount}")

price_list_2 = [1.123, 2.345, 3.456]
total_amount_2 = calculate_total_price(price_list_2)
print(f"\nOriginal prices: {price_list_2}")
print(f"Total price (rounded to 2 decimals): {total_amount_2}")
```

```python
import re

def parse_duration(s: str) -> int:
    """
    Parses a duration string (e.g., '1h30m', '45s') into total seconds.

    Args:
        s: The duration string.

    Returns:
        The total duration in seconds.

    Raises:
        ValueError: If the string format is invalid.
    """
    total_seconds = 0
    # Regular expression to find hours, minutes, and seconds.
    # It looks for one or more digits followed by 'h', 'm', or 's'.
    matches = re.findall(r'(\d+h)?(\d+m)?(\d+s)?', s)

    if not matches or not any(matches[0]): # Check if any part was matched
        raise ValueError(f"Invalid duration string format: {s}")

    # The regex always returns a tuple of 3 elements, even if empty.
    # We need to process the first (and only) actual match group.
    h_match, m_match, s_match = matches[0]

    if h_match:
        hours = int(h_match[:-1])
        total_seconds += hours * 3600
    if m_match:
        minutes = int(m_match[:-1])
        total_seconds += minutes * 60
    if s_match:
        seconds = int(s_match[:-1])
        total_seconds += seconds

    # Ensure the entire string was consumed by the regex
    # This simple check might not be robust enough for all edge cases,
    # but it catches cases like '1h30mextra'. A more complex regex
    # or string manipulation could be used for stricter validation.
    # For now, if any match was found and total_seconds is > 0, we assume valid.
    # A better approach would be to check if re.fullmatch returns something.
    if total_seconds == 0 and s != '0s' and s != '0m' and s != '0h':
         raise ValueError(f"Invalid duration string format: {s}")

    return total_seconds

# --- Demo of usage ---
print(f"'1h30m' parses to: {parse_duration('1h30m')} seconds")
print(f"'45s' parses to: {parse_duration('45s')} seconds")
print(f"'2h' parses to: {parse_duration('2h')} seconds")
print(f"'30m' parses to: {parse_duration('30m')} seconds")
print(f"'1h15m30s' parses to: {parse_duration('1h15m30s')} seconds")
print(f"'0s' parses to: {parse_duration('0s')} seconds")

# Test with invalid formats
try:
    parse_duration('invalid')
except ValueError as e:
    print(f"Error for 'invalid': {e}")

try:
    parse_duration('1h30x')
except ValueError as e:
    print(f"Error for '1h30x': {e}")

try:
    parse_duration('')
except ValueError as e:
    print(f"Error for '': {e}")
```

```python
def total(prices: list[float]) -> float:
    """
    Calculates the total of a list of prices by rounding each price to two decimal places
    first, then summing them, and finally rounding the total to two decimal places.

    Args:
        prices: A list of float values representing prices.

    Returns:
        The total sum of prices, rounded to two decimal places, after individual rounding.
    """
    rounded_prices = [round(p, 2) for p in prices]
    final_sum = sum(rounded_prices)
    return round(final_sum, 2)
```

```python
def line_total(claim) -> float:
    """
    Calculates the line total for a claim, applying a 1.18 multiplier.
    Handles cases where 'amount' might be None by treating it as 0.

    Args:
        claim: A dictionary representing a claim, expected to have an 'amount' key.

    Returns:
        The calculated line total, rounded to two decimal places.
    """
    amount = claim.get("amount", 0) # Get 'amount', default to 0 if not found or is None
    if amount is None:
        amount = 0 # Explicitly handle None values from .get() if they can occur
    return round(amount * 1.18, 2)

def report_total(claims: list[dict]) -> float:
    """
    Calculates the total of all claims in a list.

    Args:
        claims: A list of claim dictionaries.

    Returns:
        The sum of all line totals, rounded to two decimal places.
    """
    return round(sum(line_total(c) for c in claims), 2)

# --- Demo of usage ---
# Example data with a None value for amount
claims_data = [
    {"item": "Service A", "amount": 100.00},
    {"item": "Service B", "amount": None}, # This would cause the TypeError
    {"item": "Service C", "amount": 25.50}
]

print(f"Claims data: {claims_data}")
try:
    total_report_amount = report_total(claims_data)
    print(f"Total report amount: {total_report_amount}")
except TypeError as e:
    print(f"An error occurred: {e}")


# Example data with valid amounts
claims_data_valid = [
    {"item": "Product X", "amount": 50.75},
    {"item": "Product Y", "amount": 120.00}
]
print(f"\nClaims data (valid): {claims_data_valid}")
print(f"Total report amount (valid): {report_total(claims_data_valid)}")
```

```python
from collections import Counter

def get_stats(vocab):
    pairs = Counter()
    for word, freq in vocab.items():
        for i in range(len(word) - 1):
            pairs[(word[i], word[i+1])] += freq
    return pairs

def merge_pair(vocab, pair):
    new_vocab = {}
    bigram = pair[0] + pair[1]
    for word, freq in vocab.items():
        i = 0
        new_word = []
        while i < len(word):
            try:
                j = word.index(pair[0], i)
                if j < len(word) - 1 and word[j+1] == pair[1]:
                    new_word.extend(word[i:j])
                    new_word.append(bigram)
                    i = j + 2
                else:
                    new_word.append(word[i])
                    i += 1
            except ValueError:
                new_word.extend(word[i:])
                i = len(word)
        new_vocab[tuple(new_word)] = freq
    return new_vocab

corpus = 'low low low low low lower lower newest newest newest widest widest'

# every word starts as individual characters, with an end-of-word marker
vocab = {tuple(w) + ('</w>',): f for w, f in Counter(corpus.split()).items()}

for step in range(1, 9):
    stats = get_stats(vocab)
    if not stats:
        break

    best, count = stats.most_common(1)[0]
    vocab = merge_pair(vocab, best)

    print(f'merge {step}: {best[0]!r} + {best[1]!r} -> {best[0]+best[1]!r} (seen {count}x)')

print()

for k in vocab:
    print('  ', list(k))
```

```python
import time
import json # Added to parse JSON from error string
import re   # Added for regex to extract JSON from error string
from google.genai.errors import ClientError

HARD = (
    'A shop sells pens at Rs 14 each. A pack of 12 costs Rs 168. '
    'Today packs are 15% off; singles are full price. '
    'Meera needs 30 pens and wants the cheapest total. '
    'She also has a coupon for Rs 20 off any order above Rs 250, usable once. '
    'What does she pay?'
)

# Define SMART model for this cell
SMART = 'gemini-3.6-flash'

runs = []

i = 0
while i < 5:
    try:
        out = ask(
            HARD + '\n\nWork step by step. End with a line: FINAL: <number>',
            model=SMART,
            temperature=0.7,
            max_tokens=3000
        )

        line = next(
            (l for l in out.splitlines() if 'FINAL' in l.upper()),
            '(no FINAL line)'
        )

        runs.append((line.strip(), out))
        print(f'run {i+1}: {line.strip()}')
        time.sleep(0.4)
        i += 1
    except ClientError as e:
        error_message_str = str(e)
        error_dict = {}
        # Attempt to extract JSON from the error message string
        match = re.search(r'(\{.*\})', error_message_str)
        if match:
            json_str = match.group(1)
            # Replace single quotes with double quotes for valid JSON, and handle newlines
            json_str = json_str.replace("'", '"').replace('\\n', '\\\\n')
            try:
                error_dict = json.loads(json_str)
            except json.JSONDecodeError:
                pass # If parsing fails, error_dict remains empty

        error_code = error_dict.get('error', {}).get('code')
        if error_code == 429:
            retry_after = 30
            for detail in error_dict.get('error', {}).get('details', []):
                if detail.get('@type') == 'type.googleapis.com/google.rpc.RetryInfo':
                    retry_delay_str = detail.get('retryDelay', '30s')
                    retry_after = int(retry_delay_str.replace('s', ''))
                    break
            print(f"Quota exceeded for run {i+1}. Retrying in {retry_after} seconds...")
            time.sleep(retry_after + 1)
        else:
            raise

print('\ndistinct answers:', len({r[0] for r in runs}))
```

```python
REACT = """Problem: {p}\nThought:\n"""

def calculator(expression):
    """A simple calculator tool for demonstration."""
    try:
        return eval(expression)
    except Exception as e:
        return f"Error: {e}"

def react(problem, max_steps=8, verbose=True):
    t = REACT.format(p=problem)

    for _ in range(max_steps):
        out = ask(
            t,
            model=SMART,
            temperature=0,
            max_tokens=1000, # Increased max_tokens again for more complete responses
            system='''You are a reasoning agent.
            To solve the problem, first output a Thought, then either an Action or an Answer.
            Your actions must be in the format: Action: calculator("EXPRESSION").
            Your final answer must be in the format: Answer: <answer>.'''
        )

        if verbose:
            print(out)

        t += out

        if 'Answer:' in out:
            return out.split('Answer:')[-1].strip().splitlines()[0]

        # Corrected regex to match calculator('expression') or calculator("expression")
        # Using a triple-quoted raw string to avoid SyntaxError with internal quotes.
        m = re.search(r"""Action: calculator\((?:'([^']+?)'|"([^"]+?)")\)""", out)

        expr = None
        if m:
            # Try to get group 1 (single-quoted) or group 2 (double-quoted)
            expr = m.group(1) if m.group(1) else m.group(2)

        if not expr:
            return '(no action, no answer)'

        obs = calculator(expr)

        if verbose:
            print(f'Observation: {obs}\n')

        t += f'\nObservation: {obs}\n'

    return '(ran out of steps)'


print('ANSWER:', react(HARD))
```

```python
from langchain_core.runnables import RunnableParallel
import time
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

poem = (
    ChatPromptTemplate.from_template('Two line poem about {topic}')
    | llm
    | StrOutputParser()
)

joke = (
    ChatPromptTemplate.from_template('Tell me a short joke about {topic}')
    | llm
    | StrOutputParser()
)

both = RunnableParallel(joke=joke, poem=poem)

t0 = time.time()
out = both.invoke({'topic': 'space'})
par = time.time() - t0

t0 = time.time()
_ = joke.invoke({'topic': 'space'})
_ = poem.invoke({'topic': 'space'})
seq = time.time() - t0

print(out['joke'])
print(out['poem'])
print(f'\nparallel {par:.1f}s   sequential {seq:.1f}s')
```

```python
SYSTEM = (
    'Answer using ONLY the context below. '
    'If the context does not contain the answer, say exactly: '
    '"I cannot answer that from the handbook." '
    'Never use knowledge from outside the context.'
)

prompt = ChatPromptTemplate.from_messages([
    ('system', SYSTEM + '\n\nCONTEXT:\n{context}'),
    ('user', '{question}'),
])


def ask(question, k=3, show_context=False):
    hits = store.similarity_search(question, k=k)

    context = '\n---\n'.join(d.page_content for d in hits)

    if show_context:
        print('RETRIEVED:\n', context[:600], '\n')

    answer = (prompt | llm | StrOutputParser()).invoke({
        'context': context,
        'question': question
    })

    return answer, hits


a, hits = ask('How late can I submit an assignment?')
print(textwrap.fill(answer, 90))

print('\nsources:')

for d in hits:
    print(' -', d.page_content[:70].replace('\n', ' '))
```

```python
!pip -q install transformers sentencepiece

import time, warnings
warnings.filterwarnings("ignore")
from transformers import pipeline

t0 = time.time()
classifier = pipeline(
    "zero-shot-classification",
    model="MoritzLaurer/deberta-v3-base-zeroshot-v2.0"
)
classifier("warm up", candidate_labels=["a"])  # the first call is slow; get it out of the way
print(f"model ready in {time.time() - t0:.0f}s")
```

```python
# 📋 COPY THIS
ROUTES = ["billing", "technical problem", "sales question", "thank you"]

def choice(text, options, template="This message is about {}."):
    r = classifier(
        text,
        candidate_labels=options,
        hypothesis_template=template
    )
    return {
        label: round(p, 3)
        for label, p in zip(r["labels"], r["scores"])
    }
```

```python
# ✨ TRY IT
print(choice(
    "I was charged twice this month and nobody is answering.",
    ROUTES
))

print(choice(
    "The app crashes every time I open settings.",
    ROUTES
))
```

```python
def noul(text, yes, no, template="The customer is {}."):
    return choice(text, [yes, no], template=f"{template}[yes]")
```

```python
print("angry:", noul("Fix it now or I'm cancelling.", "angry or frustrated", "calm or happy"))
print("angry:", noul("Thanks, the fix worked perfectly!", "angry or frustrated", "calm or happy"))
```

```python
URGENCY = ["it can wait a week", "it should be handled today", "it needs action right now"]

def score(text, levels, template="For support, {}."):
    probs = choice(text, levels, template)
    value = sum((levels.index(level) + 1) * p for level, p in probs.items())
    return round(value, 2)
```

```python
for m in [
    "Our whole company is locked out of the dashboard since this morning, we can't work.",
    "Hi, do you offer a discount for teams of 20 people?",
    "Thanks, the fix worked perfectly!"
]:
    print(f"urgency {score(m, URGENCY)} / 3  <- {m}")
```
